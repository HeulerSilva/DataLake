```python
carros = spark.read.load("/Volumes/workspace/df_project1/data/Carros.csv",
                     format="csv", sep=";", inferSchema=True, header=True)
carros.write.mode("overwrite").saveAsTable("workspace.df_project1.carros")
```

```python
%sql
select * from workspace.df_project1.carros LIMIT 10;
```

```python
from pyspark.sql.functions import col, when
carros_df = spark.table("workspace.df_project1.carros")

carros_df.write.format("delta").save("/Volumes/workspace/df_project1/data/carros_delta")
```

```python
delta_df = spark.read.format("delta").load("/Volumes/workspace/df_project1/data/carros_delta")
delta_df.show()
```

```python
delta_df = delta_df.withColumn('TipoMotor',when(col('TipoMotor') == 0, 2).otherwise(col('TipoMotor')))
```

```python
delta_df.write.format("delta").mode("overwrite").save("/Volumes/workspace/df_project1/data/carros_delta")
updated_delta_df = spark.read.format("delta").load("/Volumes/workspace/df_project1/data/carros_delta")
updated_delta_df.show()
```
